```python
import torch
```

```python
B = 2
T = 3
D = 4

X = torch.randn(B, T, D)

print(X)
print(X.shape)
```

```python
print(X[0].shape)
print(X[0, 0].shape)
print(X[0, 0, 0].shape) #scalar
```

```python
W = torch.randn(D, D)

Y = X @ W

print("X:", X.shape)
print("W:", W.shape)
print("Y:", Y.shape)
```
